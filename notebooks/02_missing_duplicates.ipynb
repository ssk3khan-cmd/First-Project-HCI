```python
# Task 13: Count missing values
df.isnull().sum()
```

```python
# Task 14: Show only columns with missing values, sorted
missing_values = df.isnull().sum()
missing_values[missing_values > 0].sort_values(ascending=False)
```

```python
# Task 15: Missing value percentage
missing_percentage = (df.isnull().sum() / len(df) * 100).round(2)
pd.DataFrame({
    "Missing Count": df.isnull().sum(),
    "Missing Percentage": missing_percentage
}) 
```

**Student Questions (Tasks 14–15):**
1. Column with largest number of missing values: Cabin (687 missing)
2. Highest missing percentage: Cabin ≈ 77.10%
3. Age missing percentage ≈ 19.87% — significant, not small
4. Embarked missing percentage ≈ 0.22% — very minor
5. Filling Cabin would not be reliable — with over three-quarters of values missing, any imputation (mean/mode/constant) would be mostly guessed data and could seriously bias analysis. Best to leave it unchanged or drop it if unusable later.


```python
# Task 16: Remove duplicate rows
df = df.drop_duplicates().copy()
```

```python
# Task 17: Handle missing Age values with median
print("Median Age:", df["Age"].median())
df["Age"] = df["Age"].fillna(df["Age"].median())
```

```python
# Task 18: Handle missing Embarked values with mode
print("Mode of Embarked:", df["Embarked"].mode()[0])
df["Embarked"] = df["Embarked"].fillna(df["Embarked"].mode()[0])
```

```python
# Task 19: Verify missing values after cleaning
df.isnull().sum()
```
